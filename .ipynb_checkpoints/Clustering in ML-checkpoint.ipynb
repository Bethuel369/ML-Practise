Scaling is the process in which we standardise/normalise our data so that our numeric values (which ML models are trained on), are more evenly distributed and no one values adds more weight to the model than the other values

There are various ways of data scaling and their use cases depend on what your trying to achieve, some of these scalers include:

1.Standard scaler

```python
from sklearn.preprocessing import StandardScaler
import numpy as np

data=np.array([[10],[20],[30],[40],[50]])

normal_math=10-np.mean(data)/np.std(data)
print(normal_math)

print('----------------------------------')

scaler=StandardScaler()

scaled_data=scaler.fit_transform(data)

print(scaled_data)
```

2.Min Max Scaler

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

data=np.array([[10],[20],[30],[40],[50]])

scaler=MinMaxScaler()

scaled_data=scaler.fit_transform(data)

print(scaled_data)
```

3.Robust scaler

```python
from sklearn.preprocessing import RobustScaler
import numpy as np

data=np.array([[50],[60],[80],[10],[10000]])

scaler=RobustScaler()

scaled_data= scaler.fit_transform(data)

print(scaled_data)
```

4.Max Absolute Scaler

```python
from sklearn.preprocessing import MaxAbsoluteScaler

data= np.array([[-6],[-5],[4],[8],[10]])

scaler=MaxAbsoluteScaler()

scaled_data=scaler.fit_transform(data)

print(scaled_data)
```
