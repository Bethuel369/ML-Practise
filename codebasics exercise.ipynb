```python
# 1.Import necessary libraries

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.metrics import classification_report

# 2.Import data from sklearn datasets
digit_data=load_digits()

# 3.Transfer the data to a dataframe and define features & targets
df=pd.DataFrame(digit_data.data,columns=digit_data.feature_names)
df['target']=digit_data.target
df.head()

# 4. Define X an y
X = df.drop(columns=['target'])
y = df['target']

# 5.Split X and y
X_train, X_test, y_train, y_test= train_test_split(X,y)

# 6.Instate the knn model
knn=KNeighborsClassifier(n_neighbors=3)

# 7.Fit the model
knn.fit(X_train,y_train)

# 8.Predict
y_pred=knn.predict(X_test)

# 9.Measure the model's score
k_score=knn.score(X_test,y_test)
print(k_score)

# 10.Plot a confusion matrix & display
cm=confusion_matrix(y_test,y_pred)
ConfusionMatrixDisplay(confusion_matrix=cm).plot()
plt.show()

# 11.Make a classification report
print(classification_report(y_test,y_pred))
```
